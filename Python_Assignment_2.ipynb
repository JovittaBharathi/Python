**Set & Frozen Set**

```python
#Frozen Set Creation
my_set=['a','e','i','o','u','a','a','i']
```

```python
print(my_set)
```

```python
frozen_set=frozenset(my_set)
```

```python
print(frozen_set)
```

FrozenSet is unordered & immutable. It has only unique elements,doesn't allow duplicates. It has elements with different datatypes.

Hence it has only 'a','e','i','o','u' that too in unordered format in the o/p.

```python
#Changing the Value
my_set[4]="s"
```

Since FrozenSet has no Indexing(Unordered), accessing cannot be done here.
Also add,update,remove are not possible.

```python
#Set Creation
set1={1,3,5,7,9}
set2={2,3,5,8,10}
```

```python
#Intersection
set1.intersection(set2)
```
